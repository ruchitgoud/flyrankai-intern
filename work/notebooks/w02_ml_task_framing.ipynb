## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My Lane as an ML Task

I selected the Refresh / Content Opportunity Scoring lane.

The ML task is a **scoring and ranking problem**. The goal is to assign each content item an opportunity or risk score and use that score to rank pages for human review.

The system is not intended to automatically decide that a page must be refreshed. Instead, it should help prioritize which pages a reviewer should investigate first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

For the starter version, I will use `is_declining_label` as a proxy target.

The proxy is defined as:

`is_declining_label = trend_direction == "down"`

A positive example represents a content item whose observed trend direction is down.

This is only a proxy for the larger business question. It does not prove that a page will continue declining in the future or that refreshing the page will cause recovery.

For a stronger future version, I could define a future-looking target such as decline over the next 30 days using features from a previous 90-day window.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

The main success metric will be **Precision@50**.

Precision@50 measures how many of the top 50 pages selected by the model are actually positive according to the chosen target.

This metric matches the real decision because a content team has limited time and may only be able to review a fixed number of pages.

A higher Precision@50 means the ranked list is more useful for prioritizing review work.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Fixed Rule

A fixed rule can identify obvious cases using thresholds such as impressions, freshness, position, or trend.

However, multiple signals can interact in ways that are difficult to capture with one manually designed rule.

Machine learning can combine several observable signals and learn patterns associated with the target proxy. It can then produce a ranked list of pages for review.

ML is useful here only if it improves the ranking quality compared with a transparent baseline rule.

The goal is therefore not to use ML simply because it is available, but to test whether it provides better prioritization for the review decision.
